```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
```

```python

file = r'/Users/yehui/Documents/Code/Num_Analysis_Ex/课程项目2/Q1/by.csv'

data = pd.read_csv(file)

#print("描述性统计结果")
#print(data.describe())

print('检验缺失值')

print(data.info()) #检验缺失值

#print(data.head(5))
```

```python
x = data[['season','yr','mnth','holiday','weekday','workingday','weathersit','temp','atemp','hum','windspeed']]
y = data['cnt']

#设置绘图风格
plt.style.use('ggplot')
#处理中文乱码
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
#坐标轴负号的处理
plt.rcParams['axes.unicode_minus'] = False


plt.subplot(2,1,1)
plt.plot(data['instant'],data['cnt'],ls='-', lw=2, label='cosine', color='purple')

#plt.close()
plt.subplot(2,1,2)
plt.plot(data['instant'],data['windspeed'],ls='-', lw=2, label='cosine', color='purple')

# 修改x轴和y轴标签
plt.xlabel('')
plt.ylabel('气温')
# 添加图形标题
plt.title('趋势')
# 显示图形
plt.show()


x = sm.add_constant(x)
est = sm.OLS(y,x)

model = est.fit()#建立最小二乘法

print(model.summary())
```
