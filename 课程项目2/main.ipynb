```python
raw_data = pd.read_csv('共享单车使用量数据集.csv')
data_shuffle = shuffle(raw_data)
data = data_shuffle.values[:,2:]
#data = data + 1
#data[:,4] = 1
x = data[:-10,:-3]
y = data[:-10,-1]
x_test = data[-10:,:-3]
y_test = data[-10:,-1]
data
```

```python
model = LinearRegression()
model.fit(x,y)
predict = model.predict(x_test)
model.coef_
```

## 从第三个系数可以发现随着12月的到来，共享单车的热度会逐渐下降
## 从最后四个系数可以窥探出天气因素与共享单车使用量之间的正负相关性
#### 温度较高（体感温度），湿度较低，风速较小的天气更多人选择共享单车
