```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn import tree
import graphviz
import pydotplus
import random
iris = load_iris() #sepal length（萼片长度）sepal width（萼片宽度）spetal length（花瓣长度）spetal width （花瓣宽度）
target = iris.target #Setosa、Versicolour、Virginica
N = 75 #训练集样本数
data = np.zeros((N,4))
target = np.zeros((N)) 

for i in range(N):
    k = random.randint(0,149)
    data[i] = iris.data[k]
    target[i] = iris.target[k]
```

```python
clf = tree.DecisionTreeClassifier()
clf = clf.fit(data,target)
dot_data = tree.export_graphviz(clf, out_file=None)
graph = pydotplus.graph_from_dot_data(dot_data)
graph.write_pdf("mytree.pdf")

p = 0
n = 50  #测试集样本数
data_pd = np.zeros((n,4))
target_true = np.zeros((n))

for i in range(n):
    k = random.randint(0,149)
    data_pd[i] = iris.data[k]
    target_true[i] = iris.target[k]
target_pd = clf.predict(data_pd)
for i in range(n):
    if(target_true[i]==target_pd[i]):
        p = p + 1
p = p/n
print(p) #输出正确率
```
