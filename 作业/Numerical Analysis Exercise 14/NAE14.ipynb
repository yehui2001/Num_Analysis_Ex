```python
import numpy as np
import pandas as pd
from audioop import minmax
from sklearn.neural_network import MLPClassifier #加载neural network
from sklearn.model_selection import train_test_split
```

```python

Data = pd.read_csv('breast_cancer.csv')
data = np.array(Data) #转换成array操作，否则无法操作
# 拆分测试集和训练集 
X_train, X_test, Y_train, Y_test = train_test_split(data[:,0:29],data[:,30],test_size=0.25,random_state=1)  # 拆分数据集 测试集占25%
# 隐藏神经元个数确定 还有经验公式可供参考
''' 
隐藏神经元的数量应在输入层的大小和输出层的大小之间。
隐藏神经元的数量应为输入层大小的2/3加上输出层大小的2/3。
隐藏神经元的数量应小于输入层大小的两倍。
'''
#random_state控制随机状态  随机状态的作用:是控制分割数据时的影响因素，使随机分割结果可以重现。
md = MLPClassifier(solver='lbfgs', hidden_layer_sizes=[30,15], random_state=0)
md.fit(X_train,Y_train) #利用拆分数据训练 neural network

score = md.score(X_test,Y_test)  #评估模型
print(score)
```
