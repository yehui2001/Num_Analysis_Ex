```python
#导入需要使用的库
import numpy as np
import pandas as pd
import matplotlib as plt
import torch
from sklearn.datasets import load_iris
```

```python
#导入数据   0 is setosa 1 is versicolor 2  is virginica
data = np.array([[5.4,3.9,1.7,0.4,0],
[4.6,3.4,1.4,0.3,0],
[5.0,3.4,1.5,0.2,0],
[4.4,2.9,1.4,0.2,0],
[4.9,3.1,1.5,0.1,0],
[7.0,3.2,4.7,1.4,1],
[6.4,3.2,4.5,1.5,1],
[6.9,3.1,4.9,1.5,1],
[5.5,2.3,4.0,1.3,1],
[6.5,2.8,4.6,1.5,1],
[6.5,3.0,5.5,1.8,2],
[7.7,3.8,6.7,2.2,2]])
```

```python
#
x = data[:,0:-1]
x = ((x - x.min(axis=0)) / (x.max(axis=0) - x.min(axis=0))).T # 归一化处理
y = np.zeros((3,x.shape[1]))
for i in range(x.shape[1]):
    for k in range(3): #k 从0到3（不包含3）0  1 2
        if data[i,4] == k:
            y[k,i] = 1

#w = np.array(np.ones((x.shape[1],3))) #初始化权系数
w=(np.random.random([3,4])-0.5)*2
print(w)
m = x.shape[1]
```

```python
def sigmoid(x,w):
    yo = 1/(1+np.exp(w.dot(x)))
    return yo

def cost_function(x,w,y):
    yo = sigmoid(x,w)
    cost = np.sum((yo-y)**2)
    return cost/2*m

def gradient(x,w,y):
    grad = np.zeros((4,3))
    yo = sigmoid(x,w)
    print(yo)
    print(grad)
    for i in range(0,3):
        for j in range(0,4):
            grad[i,j] = -(y[i,j]-yo[i,j])*yo[i,j]*(1-yo[i,j])*(x[i,j])
    return grad

def gradiend_descent(x,w,y,eta):
    while True:
        last_w = w
        grad = gradient(x,w,y)
        w = w - eta * grad
        print(w)
        if abs(cost_function(x,last_w,y)-cost_function(x,w,y)) < 1e-15:
            break
        return w

res = gradiend_descent(x,w,y,0.0001)
```

```python

iris  = load_iris()
target = iris.target
X = iris.data
X = X[10:40,:]
target_true = target[10:40]
Y = 1/(1+np.exp(-res * X))
```
